```python
cat=[]  ## categorical
con=[]  ## continuous
for i in df.columns:
    if df[i].dtypes=="object":
      cat.append(i)
else:
    con.append(i)


cat
con
```

```python
## trate the missing values

#cat-->mode()
#con-->mean()
## fillna()

for i in df.columns:
    if df[i].dtypes=="object":
        m=df[i].mode()[0]
        df[i]=df[i].fillna(m)
    else:
        me=df[i].mean()
        df[i]=df[i].fillna(me)
```

## Bar charts - univariate categorical Variable


```python
import matplotlib.pyplot as plt
import seaborn as sns
## Univariate Bar chart foe "pipe"
plt.figure(figsize=(5,3))
sns.countplot(x='Model',data=df)
plt.title("Models")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.countplot(x='Type',data=df)
plt.title("count of vehicles types")
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x='Manufacturer',data=df)
plt.title("Manufacturer")
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x='DriveTrain',data=df)
plt.title("DriveTrain Count")
plt.show()
```

## Histogram - univariate Continuous Variable



```python
## univariate Histogram for "price"
plt.figure(figsize=(5,5))
plt.hist(x="Price",data=df,edgecolor='red')
plt.xlabel("Prices of cars")
plt.ylabel("frequency")
plt.show()
```
